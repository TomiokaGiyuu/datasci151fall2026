**Question (A)** Split a dataset into subsets

<font size = "3">

- Create a DataFrame by reading in the file "fifa23_players_basic.csv" from the quiz_data folder.

- Use ``len`` or ``.shape`` to define an integer variable called ``n``, which is the total number of rows in this DataFrame. <br>

- Use ``numpy.random.uniform`` to create a random sample of size "n" from the uniform distribution between 0 and 1. Then store these random numbers as a new column named "random_var" to the original DataFrame.
- Use ``.query`` to extract all players with ```random_var < 0.25```, and assign it to a new DataFrame called ``groupA``
- Use ``.query`` to extract all players with ```0.25 < random_var <= 0.5```, and assign it to a new DataFrame called ``groupB``.
- Use ``.query`` to extract all players with ```0.5 < random_var```, and assign it to a new DataFrame called ``groupC``.

- We expect that Group A and Group B should have roughly the same number of players. We expect that Group C is roughly twice the size of the other two groups. Verify this by calculating the ratios

$$\frac{\textrm{\# in group A}}{\textrm{\# in group B}}, \qquad \frac{\textrm{\# in group C}}{\textrm{\# in group A}}, \qquad \frac{\textrm{\# in group C}}{\textrm{\# in group B}}

````python
# Your code here
import numpy as np
import pandas as pd
fifa = pd.read_csv("./quiz_data/fifa23_players_basic.csv")
n = len(fifa)
fifa["random_var"] = np.random.uniform(0, 1, n)
groupA = fifa.query("random_var < 0.25")
groupB = fifa.query("random_var >0.25 and random_var <= 0.5")
groupC = fifa.query("random_var > 0.5")
print(len(groupA), len(groupB), len(groupC))
print(len(groupA)/len(groupB), len(groupB)/len(groupC), len(groupC)/len(groupA))
    
````

**Question (B)** Merging DataFrames

<font size = "3">

In the code cell below, a Pandas DataFrame is created manually that describes the **engine type** associated with each number of cylinders found in the "features.csv" dataset. It allows us to determine the engine type of a given car.

````python
import pandas as pd

cylinders_lst   = [3, 4, 5, 6, 8]
engine_type_lst = ["Inline-3", "Inline-4", "Inline-5", "V6", "V8"]

df_engine = pd.DataFrame({"cylinders": cylinders_lst, "engine_type": engine_type_lst})

df_engine
````

- This dataset should be familiar to you by now. It contains information about the fuel efficiency ("mpg"), number of cylinders, weight, and other features of various cars.

- Perform the merge so that the resulting DataFrame has 5 columns in the following order: "engine_type", "cylinders", "mpg", "weight", "vehicle_id".

- In addition, perform the merge so that the ordering of the rows matches the ordering of "features.csv"

````python
# Your code here
features = pd.read_csv("./quiz_data/features.csv")
df_engine_reorder = df_engine[["engine_type", "cylinders"]]
features_select = features[["cylinders", "mpg", "weight", "vehicle_id"]]
features_merged = pd.merge(left = df_engine_reorder, right = features_select, on = "cylinders", how = "left")
print(features_merged)
````

**Question (D)** Transforming variables

<font size = "3">

In a chemistry lab, students prepared samples of water to be brought to a boil. For each sample, they recorded the mass of the water and its starting temperature, using a thermometer that reads in degrees Fahrenheit.

Masses were recorded in **grams (g)**, and temperatures were recorded in **degrees Fahrenheit (°F)**.

1. Read in the file “water_heating.csv” from the quiz_data folder and save it as a DataFrame.

2. Using the `.apply` method, define a Pandas Series that contains the starting temperature in degrees Celsius, using this formula:

$$ \textrm{temperature in °C} = \frac{5}{9} \cdot (\textrm{temperature in °F} - 32)$$

3. Add a new column to the DataFrame named “Heat to Boil (J)” that contains the heat energy needed to bring each sample to its boiling point of 100 °C. The formula for heat energy is given below, where $m$ is the mass in grams, $c = 4.184$ is the specific heat of water (in units of J/(g·°C)), and $T$ is the starting temperature **in degrees Celsius**.

$$ \textrm{heat energy} = m \cdot c \cdot (100 - T)$$


````python
# Your code here
water = pd.read_csv("./quiz_data/water_heating.csv")
def FtoC(F):
    return 5/9*(F-32)
#print(water.columns)
TempC = water["Temp (F)"].apply(FtoC)
water["Heat to Boil (J)"] = water["Mass (g)"]*4.184*(100-TempC)
water.head()
````

**Question (E)** More merging

Read in the files "sprint_results.csv" and "constructors.csv" from the quiz_data folder, assigning them to two different DataFrames.

Use `pandas.merge` to create a new DataFrame called `df_merged` by combining the two DataFrames you created by reading in the data:

1. The only column they have in common is "constructorId".

2. The DataFrame `df_merged` should have exactly 6 columns in the following order:

    - "resultId", "raceId", "constructorId", "points" (taken from the "sprint_results" DataFrame)

    - "constructorRef", "nationality" (taken from the "constructors" DataFrame)

3. The order of the rows should match the ordering of "sprint_results.csv".

4. You only need a single call to `pd.merge` for this problem. Only merge the required columns (if you know how to use the `.drop` method, don't use it after merging the DataFrames!)

5. Using the `.query` method on `df_merged`, extract the rows where the constructor's nationality is "British" **and** the number of points is **strictly greater than** 0. Assign the result to a DataFrame called `df_british`.

6. Using ``.shape``, define an integer variable `num_british` which is the number of rows in `df_british`.

````python
# Your code here
sprint = pd.read_csv("./quiz_data/sprint_results.csv")
constructors = pd.read_csv("./quiz_data/constructors.csv")
sprint_select = sprint.iloc[:, [0, 1, 3, 9]]
constructors_select = constructors.iloc[:, [0, 1, 3]]
df_merged = pd.merge(left = sprint_select, right = constructors_select, on = "constructorId", how = "left")
print(df_merged.head())

df_british = df_merged.query("points > 0 and nationality=='British'")
num_british = df_british.shape[0]
print(df_british)
print(num_british)
````
